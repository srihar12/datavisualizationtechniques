```python

import pandas as pd
df = pd.read_csv(r"C:\Users\kalap\Downloads\McDonalds_financial_statements.csv")
mcd = df.copy()
mcd.drop_duplicates(inplace=True)
print("Missing Values:")
print(mcd.isnull().sum())
mcd.columns = (
    mcd.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("$", "", regex=False)
)
print("Original Shape:", df.shape)
print("Cleaned Shape:", mcd.shape)
display(mcd.head())

```

```python
# Revenue Growth
mcd["revenue_growth"] = mcd["revenue_b"].pct_change() * 100

# Earnings Growth
mcd["earnings_growth"] = mcd["earnings_b"].pct_change() * 100

# Debt to Asset Ratio
mcd["debt_to_asset_ratio"] = (
    mcd["total_debt_b"] / mcd["total_assets_b"]
) * 100

# Liability to Asset Ratio
mcd["liability_to_asset_ratio"] = (
    mcd["total_liabilities_b"] / mcd["total_assets_b"]
) * 100

# Display result
display(mcd.head())
```
