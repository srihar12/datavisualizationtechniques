```python
import pandas as pd
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("belbino/indian-banking-transactions-20192024")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv("indian_banking_transactions.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
# 2. CHECK DATA
# ============================================

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ============================================
# 3. MAKE A COPY FOR CLEANING
# ============================================

bank = df.copy()


# ============================================
# 4. HANDLE MISSING VALUES
# ============================================

bank["customer_id"] = bank["customer_id"].fillna("Unknown")
bank["transaction_type"] = bank["transaction_type"].fillna("Unknown")
bank["account_type"] = bank["account_type"].fillna("Unknown")
bank["merchant_category"] = bank["merchant_category"].fillna("Unknown")
bank["transaction_direction"] = bank["transaction_direction"].fillna("Unknown")

bank["transaction_amount"] = bank["transaction_amount"].fillna(0)
bank["account_balance"] = bank["account_balance"].fillna(0)


# ============================================
# 5. REMOVE DUPLICATES
# ============================================

bank.drop_duplicates(inplace=True)


# ============================================
# 6. CONVERT DATE
# ============================================

bank["transaction_date"] = pd.to_datetime(
    bank["transaction_date"],
    errors="coerce"
)


# ============================================
# 7. CLEAN COLUMN NAMES
# ============================================

bank.columns = (
    bank.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)


# ============================================
# 8. FINAL CHECK
# ============================================

print("Original rows:", len(df))
print("Cleaned rows:", len(bank))
print("Columns:", len(bank.columns))

print("\nMissing values after cleaning:")
print(bank.isnull().sum()[bank.isnull().sum() > 0])

print("\nDuplicate rows after cleaning:")
print(bank.duplicated().sum())

print("\nData types:")
print(bank.dtypes)

bank.head()
```

```python
# 9. CONNECT TO MYSQL
# ============================================

from sqlalchemy import create_engine, text

username = "root"
password = "2129"
host = "localhost"
port = 3306
```

```python
# 10. CREATE DATABASE
# ============================================

server_engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}"
)

with server_engine.connect() as connection:
    connection.execute(
        text("CREATE DATABASE IF NOT EXISTS banking_analytics")
    )
```

```python
# 11. CONNECT TO DATABASE
# ============================================

db_engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/banking_analytics"
)
```

```python
# 12. LOAD DATA INTO MYSQL
# ============================================

bank.to_sql(
    "banking_transactions",
    con=db_engine,
    if_exists="replace",
    index=False,
    chunksize=5000
)
```

```python
# 13. VERIFY MYSQL DATA
# ============================================

print(
    pd.read_sql(
        "SELECT COUNT(*) AS total_rows FROM banking_transactions",
        db_engine
    )
)

print(
    pd.read_sql(
        "SELECT * FROM banking_transactions LIMIT 5",
        db_engine
    )
)
```
